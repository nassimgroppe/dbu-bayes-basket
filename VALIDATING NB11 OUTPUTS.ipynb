```python
import pandas as pd
from pathlib import Path

PROC = Path("/Users/lisagroppe/Desktop/data/processed")

# Welche Dateien liegen vor?
print("Dateien in data/processed/:")
for f in sorted(PROC.iterdir()):
    size_mb = f.stat().st_size / 1024 / 1024
    print(f"  {f.name:40s} {size_mb:8.2f} MB")
```

```python
# basket_features.parquet einlesen
basket = pd.read_parquet(PROC / "basket_features.parquet")

print(f"Shape:    {basket.shape}")
print(f"Aktien:   {basket['Ticker'].nunique()}")
print(f"Zeitraum: {basket['Date'].min()} bis {basket['Date'].max()}")
```

```python
print("Spalten:")
for col in basket.columns:
    print(f"  - {col} ({basket[col].dtype})")
```
